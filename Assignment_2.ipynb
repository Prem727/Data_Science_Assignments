1.	Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.


```python
list_int = [1, 2, 3, 4, 5]

# Append
list_int.append(6)
print("After append:", list_int)
# extend
list_int2 = [7, 8, 9]
list_int.extend(list_int2)
print("After extend:", list_int)
# insert
list_int.insert(0, 0)
print("After insert:", list_int)
#remove
list_int.remove(0)
print("After remove:", list_int)
#pop
popped_element = list_int.pop()
print("Popped element:", popped_element)
print("After pop:", list_int)
#index
index_of_5 = list_int.index(5)
print("Index of 5:", index_of_5)
#count
count_of_3 = list_int.count(3)
print("Count of 3:", count_of_3)
#sort
list_int.sort()
print("After sort:", list_int)
#reverse
list_int.reverse()
print("After reverse:", list_int)
#clear
list_int.clear()
print("After clear:", list_int)
```

# **Section C: Exploring Pandas**
11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows

```python
import pandas as pd

# 1. Create a small DataFrame manually with 10 rows
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Henry', 'Alice', 'Ivy'],
    'Age': [25, 30, 22, 28, 35, 27, 32, 29, 25, 31],
    'City': ['New York', 'Paris', 'London', 'Tokyo', 'Berlin', 'Sydney', 'Toronto', 'Boston', 'New York', 'Chicago'],
    'Salary': [70000, 85000, np.nan, 120000, 60000, np.nan, 95000, 80000, 70000, 90000]
}

df = pd.DataFrame(data)
print("--- Original DataFrame ---")
print(df)

# 2. Use .info() and .describe()
print("\n--- DataFrame Info ---")
df.info()

print("\n--- DataFrame Descriptive Statistics ---")
print(df.describe(include='all'))

# 3. Select 'Name' and 'City' columns
name_city_df = df[['Name', 'City']]
print("\n--- Selected 'Name' and 'City' Columns ---")
print(name_city_df)

# 4. Drop 'City' column
df_dropped = df.drop(columns=['City'])
print("\n--- DataFrame after dropping 'City' column ---")
print(df_dropped)

# 5. Fill any missing values in 'Salary' column with the mean
mean_salary = df_dropped['Salary'].mean()
df_dropped['Salary'] = df_dropped['Salary'].fillna(mean_salary)
print(f"\n--- Filled missing Salaries with mean ({mean_salary:.2f}) ---")
print(df_dropped)

# 6. Remove any duplicate rows
df_final = df_dropped.drop_duplicates()
print("\n--- Final DataFrame after removing duplicate rows ---")
print(df_final)
```
