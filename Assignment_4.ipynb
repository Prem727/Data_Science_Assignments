```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
patient_df = pd.read_csv("Patient_Data.csv")

print(patient_df)

billing_df = pd.read_csv("Billing_Data.csv")

print(billing_df)
```

# **Task**

```python
#Task-1
#Load the patient dataset and show summary with info().
print("Patient Dataset Information:")
patient_df.info()
```

```python
#Task-2
#Select Billing-Related Columns
billing_columns = patient_df[
    ['PatientID', 'Department', 'Doctor', 'BillAmount']
]

print("Billing Related Columns:")
print(billing_columns)
```

```python
#Task-3
#Drop Administrative Columns
patient_cleaned = patient_df.drop(
    columns=['ReceptionistID', 'CheckInTime']
)

print("After dropping administrative columns:")
print(patient_cleaned)
```

```python
#Task-4
#Total Bill Amount per Department
department_bill = patient_df.groupby(
    'Department'
)['BillAmount'].sum()

print("Total Bill Amount per Department:")
print(department_bill)
```

```python
#task-5
#Remove Duplicate Patients
patient_unique = patient_cleaned.drop_duplicates(
    subset=['PatientID']
)

print("After removing duplicate patients:")
print(patient_unique)
```

```python
#task-6
#Fill Missing BillAmount
mean_bill = patient_unique['BillAmount'].mean()

print("Mean Bill Amount:", mean_bill)

patient_unique['BillAmount'] = patient_unique['BillAmount'].fillna(
    mean_bill
)

print("After filling missing BillAmount:")
print(patient_unique)

print("Missing values after cleaning:")
print(patient_unique.isnull().sum())
```

```python
#task-7
#Merge Billing Dataset with Patient Dataset
merged_df = pd.merge(
    patient_unique,
    billing_df,
    on='PatientID',
    how='inner'
)

print("Merged Patient and Billing Dataset:")
print(merged_df)
```

```python
#task-8
#Create New Patients DataFrame

new_patients = pd.DataFrame({
    'PatientID': [106, 107],
    'Name': ['Frank', 'Grace'],
    'Department': ['Cardiology', 'Neurology'],
    'Doctor': ['Dr. Smith', 'Dr. John'],
    'BillAmount': [6800, 5500]
})

print("New Patients:")
print(new_patients)

patient_with_new = pd.concat(
    [patient_unique, new_patients],
    ignore_index=True
)

print("Patient Dataset After Adding New Patients:")
print(patient_with_new)
```

```python
#task-9
#Create New Billing Category Columns
billing_categories = billing_df[
    ['InsuranceCovered', 'FinalAmount']
]

print("New Billing Category Columns:")
print(billing_categories)

patient_billing_columns = pd.concat(
    [
        patient_unique.reset_index(drop=True),
        billing_categories.reset_index(drop=True)
    ],
    axis=1
)

print("Column-wise Concatenation:")
print(patient_billing_columns)
```

```python
#Final Cleaned Data Set
final_df = merged_df.copy()

print("Final Cleaned Dataset:")
print(final_df)

print("Final Dataset Information:")
final_df.info()

print("\nMissing Values:")
print(final_df.isnull().sum())

print("\nDuplicate Patient IDs:")
print(final_df['PatientID'].duplicated().sum())
```

```python
#Further Analysis
department_revenue = final_df.groupby(
    'Department'
)['FinalAmount'].sum()

print("Revenue by Department:")
print(department_revenue)

#doctor performance
doctor_revenue = final_df.groupby(
    'Doctor'
)['FinalAmount'].sum()

print("\nRevenue by Doctor:")
print(doctor_revenue)
```

# **Final Conclusion**
The hospital patient and billing datasets were successfully analyzed and cleaned using pandas. The patient dataset was inspected using info(), and the required billing-related columns were selected. Administrative columns were removed, duplicate patient records were eliminated based on PatientID, and missing BillAmount values were replaced with the mean bill amount. The patient and billing datasets were merged using PatientID. Additional patient records were concatenated row-wise, while billing category columns were concatenated column-wise. The final cleaned dataset contains accurate patient and billing information and can be used for further analysis such as department-wise revenue and doctor performance.
