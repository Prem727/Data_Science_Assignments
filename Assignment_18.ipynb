```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("/content/sonardataset.csv")

print("Dataset Shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nTarget Distribution:")
print(df.iloc[:, -1].value_counts())

print("\nStatistical Summary:")
display(df.describe())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Separate features and target
X = df.iloc[:, :-1].copy()
y = df.iloc[:, -1].copy()

# Convert all features to numeric
X = X.apply(pd.to_numeric, errors="coerce")

# Handle missing values using median
X = X.fillna(X.median())

# Encode target
y = y.map({
    "M": 1,
    "R": 0
})

print("Missing values after preprocessing:")
print(X.isnull().sum().sum())

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Standardization
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Training shape:", X_train_scaled.shape)
print("Testing shape:", X_test_scaled.shape)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

# Set random seed
tf.random.set_seed(42)
np.random.seed(42)

# Create basic ANN
basic_model = Sequential([
    Dense(32, activation="relu", input_shape=(60,)),
    Dense(16, activation="relu"),
    Dense(1, activation="sigmoid")
])

# Compile model
basic_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# Display architecture
basic_model.summary()

# Train model
history_basic = basic_model.fit(
    X_train_scaled,
    y_train,
    validation_split=0.20,
    epochs=50,
    batch_size=16,
    verbose=0
)

print("Basic ANN training completed.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Predictions
basic_prob = basic_model.predict(
    X_test_scaled,
    verbose=0
).ravel()

basic_pred = (
    basic_prob >= 0.5
).astype(int)

# Metrics
basic_accuracy = accuracy_score(
    y_test,
    basic_pred
)

basic_precision = precision_score(
    y_test,
    basic_pred
)

basic_recall = recall_score(
    y_test,
    basic_pred
)

basic_f1 = f1_score(
    y_test,
    basic_pred
)

print("Basic ANN Performance")
print("---------------------")
print("Accuracy :", round(basic_accuracy, 4))
print("Precision:", round(basic_precision, 4))
print("Recall   :", round(basic_recall, 4))
print("F1-Score :", round(basic_f1, 4))

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        basic_pred,
        target_names=["Rock", "Mine"]
    )
)

# Training history
plt.figure(figsize=(12, 5))

plt.plot(
    history_basic.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history_basic.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("Basic ANN Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()

plt.show()

# Confusion matrix
plt.figure(figsize=(6, 5))

sns.heatmap(
    confusion_matrix(y_test, basic_pred),
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Rock", "Mine"],
    yticklabels=["Rock", "Mine"]
)

plt.title("Basic ANN Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
from tensorflow.keras.optimizers import Adam

# Function to build ANN
def build_ann(
    hidden_layers=(32, 16),
    activation="relu",
    learning_rate=0.001
):

    model = Sequential()

    # First hidden layer
    model.add(
        Dense(
            hidden_layers[0],
            activation=activation,
            input_shape=(60,)
        )
    )

    # Additional hidden layers
    for units in hidden_layers[1:]:
        model.add(
            Dense(
                units,
                activation=activation
            )
        )

    # Output layer
    model.add(
        Dense(
            1,
            activation="sigmoid"
        )
    )

    model.compile(
        optimizer=Adam(
            learning_rate=learning_rate
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model


# Hyperparameter configurations
configs = [
    {
        "name": "Model 1",
        "layers": (32, 16),
        "activation": "relu",
        "lr": 0.001
    },
    {
        "name": "Model 2",
        "layers": (64, 32),
        "activation": "relu",
        "lr": 0.001
    },
    {
        "name": "Model 3",
        "layers": (64, 32, 16),
        "activation": "relu",
        "lr": 0.001
    },
    {
        "name": "Model 4",
        "layers": (64, 32),
        "activation": "tanh",
        "lr": 0.001
    },
    {
        "name": "Model 5",
        "layers": (64, 32),
        "activation": "relu",
        "lr": 0.0005
    }
]

results = []

best_model = None
best_f1 = -1
best_config = None

for config in configs:

    model = build_ann(
        hidden_layers=config["layers"],
        activation=config["activation"],
        learning_rate=config["lr"]
    )

    model.fit(
        X_train_scaled,
        y_train,
        validation_split=0.20,
        epochs=50,
        batch_size=16,
        verbose=0
    )

    probabilities = model.predict(
        X_test_scaled,
        verbose=0
    ).ravel()

    predictions = (
        probabilities >= 0.5
    ).astype(int)

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        zero_division=0
    )

    results.append({
        "Model": config["name"],
        "Layers": str(config["layers"]),
        "Activation": config["activation"],
        "Learning Rate": config["lr"],
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    })

    if f1 > best_f1:
        best_f1 = f1
        best_model = model
        best_config = config


results_df = pd.DataFrame(results)

display(
    results_df.sort_values(
        "F1-Score",
        ascending=False
    )
)

print("\nBest Configuration:")
print(best_config)
```

```python
# Best model predictions
tuned_prob = best_model.predict(
    X_test_scaled,
    verbose=0
).ravel()

tuned_pred = (
    tuned_prob >= 0.5
).astype(int)

# Tuned metrics
tuned_accuracy = accuracy_score(
    y_test,
    tuned_pred
)

tuned_precision = precision_score(
    y_test,
    tuned_pred,
    zero_division=0
)

tuned_recall = recall_score(
    y_test,
    tuned_pred,
    zero_division=0
)

tuned_f1 = f1_score(
    y_test,
    tuned_pred,
    zero_division=0
)

print("Tuned ANN Performance")
print("---------------------")
print("Accuracy :", round(tuned_accuracy, 4))
print("Precision:", round(tuned_precision, 4))
print("Recall   :", round(tuned_recall, 4))
print("F1-Score :", round(tuned_f1, 4))

# Comparison
comparison = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score"
    ],
    "Basic ANN": [
        basic_accuracy,
        basic_precision,
        basic_recall,
        basic_f1
    ],
    "Tuned ANN": [
        tuned_accuracy,
        tuned_precision,
        tuned_recall,
        tuned_f1
    ]
})

display(
    comparison.round(4)
)

# Confusion matrix
plt.figure(figsize=(6, 5))

sns.heatmap(
    confusion_matrix(y_test, tuned_pred),
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Rock", "Mine"],
    yticklabels=["Rock", "Mine"]
)

plt.title("Tuned ANN Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()

print("\nTuned Model Classification Report:")
print(
    classification_report(
        y_test,
        tuned_pred,
        target_names=["Rock", "Mine"]
    )
)
```

# Conclusion

An Artificial Neural Network was developed to classify sonar signals as
either Mines or Rocks.

The dataset contains 208 observations and 60 numerical sonar features.
Missing values were checked and numerical features were standardized before
training the neural network.

A basic ANN containing hidden layers with ReLU activation was first trained.
Its performance was evaluated using accuracy, precision, recall, and F1-score.

Hyperparameter tuning was then performed by experimenting with different:

- Number of hidden layers
- Number of neurons
- Activation functions
- Learning rates

The tuned ANN was compared with the basic ANN using the same testing data.

The final results show whether hyperparameter tuning improved the model's
ability to distinguish mines from rocks.

### Practical Interpretation

In a real-world underwater sonar application, correctly identifying a mine
is particularly important because missing a mine could create a serious
safety risk. Therefore, recall for the Mine class should receive particular
attention in addition to overall accuracy.

### Limitations

The dataset is relatively small, containing only 208 observations. Neural
networks can overfit small datasets, so validation techniques and
regularization can be useful.

The original study also used carefully controlled training and testing
splits based on sonar aspect angles. A simple random train-test split may
not reproduce those exact experimental conditions.

### Future Improvements

Future work could include:

1. Cross-validation.
2. Dropout and L2 regularization.
3. Early stopping.
4. More systematic hyperparameter optimization.
5. Comparing ANN performance with SVM, Random Forest, and other classifiers.
6. Using the original aspect-angle-dependent experimental methodology.

### Final Result

The tuned ANN should be selected based on the actual test-set metrics
displayed in the notebook rather than assuming that a particular
architecture will always perform best.
