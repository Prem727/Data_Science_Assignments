```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Cardiotocographic.csv")

print(df)

print("Shape of Dataset:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
print(df.head())

print("Dataset Information:")
df.info()
```

# **Task**

```python
#Missing Values
print("Missing Values:")
print(df.isnull().sum())
```

```python
#Duplicate Values
print("Number of Duplicate Rows:")
print(df.duplicated().sum())

#remove
df = df.drop_duplicates()

print("Shape after removing duplicates:")
print(df.shape)

#data types
print("Data Types:")
print(df.dtypes)

print("Numerical Columns:")
print(df.select_dtypes(include=np.number).columns.tolist())
```

```python
#Handle Missing Values
numerical_columns = df.select_dtypes(include=np.number).columns

for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())

print("Missing Values After Imputation:")
print(df.isnull().sum())
```

```python
#Cleaned Data Set
print("Final Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

#Statistical Summary
print(df.describe())
```

```python
#Statistical Table
statistics = pd.DataFrame({
    'Mean': df.mean(),
    'Median': df.median(),
    'Standard Deviation': df.std(),
    'IQR': df.quantile(0.75) - df.quantile(0.25)
})

print(statistics)
```

**Statistical Summary — Observations**

The statistical summary provides information about the central tendency and variability of the numerical variables. The mean and median help identify the typical values, while the standard deviation and interquartile range describe the spread of the observations. Some variables show a noticeable difference between their mean and median, suggesting skewness and the possible presence of extreme observations. Variables with larger standard deviations have greater variability compared with variables with smaller standard deviations.

```python
#Histograms
for column in df.columns:
    plt.figure(figsize=(8, 5))
    plt.hist(df[column], bins=30)
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.show()
```

```python
#Boxplots
for column in df.columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[column])
    plt.title(f"Box Plot of {column}")
    plt.xlabel(column)
    plt.show()
```

```python
#Outlier Detection
outlier_counts = {}

for column in df.columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_counts[column] = len(outliers)

print("Number of Potential Outliers:")
print(pd.Series(outlier_counts))
```

```python
#Correlation Matrix
correlation = df.corr()

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
#Find Strong Correlations
correlation_pairs = correlation.unstack()

correlation_pairs = correlation_pairs[
    correlation_pairs.index.get_level_values(0) !=
    correlation_pairs.index.get_level_values(1)
]

correlation_pairs = correlation_pairs.sort_values(
    key=abs,
    ascending=False
)

print(correlation_pairs.head(10))
```

```python
#Scatter Plot
plt.figure(figsize=(8, 5))

plt.scatter(df['LB'], df['ASTV'])

plt.title("Baseline Fetal Heart Rate vs Abnormal Short-Term Variability")
plt.xlabel("LB")
plt.ylabel("ASTV")

plt.show()
```

```python
#Pair Plot
selected_columns = [
    'LB',
    'AC',
    'FM',
    'UC',
    'ASTV',
    'MSTV'
]

sns.pairplot(df[selected_columns])

plt.show()
```

```python
#Violin Plot
plt.figure(figsize=(12, 6))

sns.violinplot(
    data=df[
        ['LB', 'ASTV', 'MSTV', 'ALTV', 'MLTV']
    ]
)

plt.title("Distribution of Selected Variables")
plt.xticks(rotation=45)
plt.show()
```

```python
#NSP Frequency
print("NSP Value Counts:")
print(df['NSP'].value_counts())

#bar chart
plt.figure(figsize=(8, 5))

df['NSP'].value_counts().sort_index().plot(kind='bar')

plt.title("Frequency of NSP Categories")
plt.xlabel("NSP")
plt.ylabel("Frequency")
plt.xticks(rotation=0)

plt.show()
```

**NSP Interpretation:**
NSP Category Analysis

The NSP variable represents the classification of the fetal condition. The bar chart shows the frequency of observations in each category. Comparing the category frequencies helps identify whether the dataset is balanced or whether some fetal condition categories are represented more frequently than others.

**Temporal Analysis**

The dataset does not contain an explicit date or time variable. Therefore, a time-series trend analysis cannot be performed. The analysis instead focuses on relationships, distributions, variability, and patterns among the available numerical variables.

**Pattern Recognition and Insights**

The correlation analysis helps identify relationships among the cardiotocographic variables. Some variables may show positive or negative associations, while others have weak relationships. The histograms and boxplots reveal differences in distributions, variability, skewness, and potential outliers. The pair plot provides additional visual evidence of relationships between selected variables. The NSP frequency analysis also shows how the observations are distributed across the fetal condition categories.
