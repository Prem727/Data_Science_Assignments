# **Exercise 1: Prime Numbers**

```python
num = int(input("Enter a number: "))

if num <= 1:
    print("Not Prime")

else:
    is_prime = True

    for i in range(2, int(num**0.5)+1):

        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print("Prime")
    else:
        print("Not Prime")
```

# **Exercise 2: Product of Random Numbers**

```python
import random

# Generate two random numbers
num1 = random.randint(1, 10)
num2 = random.randint(1, 10)

# Ask the user for the product
answer = int(input(f"What is {num1} × {num2}? "))

# Check the answer
if answer == num1 * num2:
    print("Correct! Well done.")
else:
    print(f"Incorrect. The correct answer is {num1 * num2}.")
```

# **Exercise 3: Squares of Even/Odd Numbers**
Create a Python script that prints the squares of all even or odd numbers within the range of 100 to 200. Choose either even or odd numbers and document your choice in the code.


```python
#squares of odd numbers from 100 to 200

for num in range(100, 201):
    if num % 2 != 0:
        print(f"{num}^2 = {num ** 2}")


# Printing squares of even numbers from 100 to 200

for num in range(100, 201):
    if num % 2 == 0:  # Check if the number is even
        print(f"{num}^2 = {num ** 2}")
```

# **Exercise 4:**
1. Problem Statement (Question): Hospital Billing System
2. A hospital charges its patients based on the following:
*	Room charges per day: ₹2,000
*	Doctor consultation fee (flat): ₹1,500
*	Lab test charges: Based on the number of tests taken. Each test costs ₹300
*	Medicine charges: Total cost of medicines provided
*	Discount: If the total bill (before discount) exceeds ₹10,000, a 10% discount is applied
3. Write a Python program to calculate the final bill for a patient who:
*	Stayed for 4 days
* Had 3 lab tests
*	Had medicine charges of ₹2,400



```python
# Hospital Billing System

room_charges_per_day = 2000
doctor_fee = 1500
test_charge_per_test = 300

# Patient details (i have given for multiple user inputs)
days_stayed = int(input("Enter number of days stayed: "))
num_tests = int(input("Enter number of lab tests: "))
medicine_charges = float(input("Enter medicine charges: "))

# Calculations
room_charges = days_stayed * room_charges_per_day
lab_charges = num_tests * test_charge_per_test

total_bill = room_charges + doctor_fee + lab_charges + medicine_charges

# Apply discount if bill exceeds ₹10,000
if total_bill > 10000:
    discount = total_bill * 0.10
else:
    discount = 0

final_bill = total_bill - discount

# Display bill
print("Room Charges:", room_charges)
print("Doctor Fee:", doctor_fee)
print("Lab Charges:", lab_charges)
print("Medicine Charges:", medicine_charges)
print("Total Bill:", total_bill)
print("Discount:", discount)
print("Final Bill:", final_bill)
```
